```python
# Лабораторная работа №3
# Ермакова Арина ЦИБ-251

## Задание: Анализ эффективности (Счетчик обменов)
# Напишите функцию bubble_sort_count_swaps(arr), которая сортирует массив
# пузырьком и возвращает количество перестановок (swaps), совершенных во
# время сортировки. Массив должен быть отсортирован на месте.


def bubble_sort_count_swaps(arr):
    swaps = 0
    n = len(arr)

    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:

                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swaps += 1
                swapped = True

        if not swapped:
            break

    return swaps
```

```python
print(bubble_sort_count_swaps([3, 2, 1]))
print(bubble_sort_count_swaps([1, 2, 3]))
```
